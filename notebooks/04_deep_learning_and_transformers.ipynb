## 1. Feedforward Multi-Layer Perceptron (MLP)


```python
import numpy as np
mlp = TextMLPClassifier(input_dim=50, hidden_dims=[32, 16], num_classes=4)
dummy_X = np.random.randn(80, 50).astype("float32")
dummy_y = np.random.randint(0, 4, size=80)
losses = mlp.fit(dummy_X, dummy_y, epochs=3)
print("Training Loss Progression:", losses)
```

## 2. Vanishing Gradient Diagnostic: Vanilla RNN vs. LSTM


```python
seq_rnn = SequenceRNN(vocab_size=1000, hidden_dim=64, num_classes=4)
diag = seq_rnn.diagnose_vanishing_gradient(sequence_length=50)
print("Vanishing Gradient Diagnostic:")
print("  Vanilla RNN Gradient Norm :", diag["vanilla_rnn_grad_norm"])
print("  LSTM Gradient Norm        :", diag["lstm_grad_norm"])
if "ratio_lstm_to_rnn" in diag:
    print("  Gradient Preservation     :", f"{diag['ratio_lstm_to_rnn']:.2f}x")
```
