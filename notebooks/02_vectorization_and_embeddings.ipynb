# Text Representation Techniques

In this notebook, we explore:
1. Vector Space Models: Bag-of-Words (BoW) & Sparsity Analysis
2. Mathematical Term Frequency - Inverse Document Frequency (TF-IDF)
3. Distributed Embeddings: Custom Word2Vec (CBOW vs. Skip-gram) and GloVe
4. Contextual Embeddings with Transformers (BERT Polysemy Analysis)


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from nlp_engine.representation.bow import BagOfWords
from nlp_engine.representation.tfidf import TFIDFModel
from nlp_engine.representation.embeddings import Word2VecEngine
from nlp_engine.representation.contextual import ContextualEmbedder
```

## 1. Bag of Words Vectorization & Sparsity


```python
corpus = [
    "invoice charged twice for order ORD-101",
    "system crash throwing error 503 dashboard lag",
    "package delivery delayed for shipment tracking",
    "please refund excess subscription fee credit card"
]

bow = BagOfWords()
bow_mat = bow.fit_transform(corpus)
print("BoW Matrix Shape:", bow_mat.shape)
print("Matrix Sparsity:  ", f"{bow.calculate_sparsity(bow_mat):.2f}%")
print("Vocabulary Sample:", bow.feature_names_[:8])
```

## 2. Mathematical TF-IDF Representation


```python
tfidf = TFIDFModel(smooth_idf=True)
tfidf_mat = tfidf.fit_transform(corpus)
print("Top TF-IDF Keywords for Document 0:")
print(tfidf.get_top_keywords(corpus[0], top_k=3))
```

## 3. Distributed Semantic Word2Vec Embeddings


```python
tokenized = [doc.split() for doc in corpus]
w2v = Word2VecEngine(vector_size=50, window=3, min_count=1, sg=1)
w2v.fit(tokenized)
print("Cosine Similarity ('invoice', 'charged'):", w2v.similarity("invoice", "charged"))
```

## 4. Contextual Embeddings & Polysemy (BERT)


```python
ctx = ContextualEmbedder()
poly_res = ctx.demonstrate_polysemy(
    "charge",
    "Please refund the unexpected credit card charge.",
    "Connect the cable to charge the phone battery."
)
print("Polysemy Analysis Result:", poly_res)
```
