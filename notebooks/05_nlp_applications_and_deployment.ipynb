```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from nlp_engine.data.generator import generate_support_tickets
from nlp_engine.pipeline.orchestrator import NLPOrchestrator
from nlp_engine.applications.translation import SupportTranslator
```

## 1. End-to-End Orchestrator Ingestion & Training


```python
df = generate_support_tickets(n_samples=200, random_state=42)
orchestrator = NLPOrchestrator()
orchestrator.fit(df)
```

## 2. Multi-Stage Ticket Triage & Analytics


```python
sample_ticket = "URGENT: Billed $49 twice for order ORD-99214 on 2026-04-12 for Pro Cloud Subscription. Please refund immediately!"
result = orchestrator.process_ticket(sample_ticket)

import json
print(json.dumps(result, indent=2))
```

## 3. Translation & BLEU Score Evaluation


```python
translator = SupportTranslator()
eval_res = translator.evaluate_translation(
    reference="please refund my order immediately",
    hypothesis="please refund my order now"
)
print("Translation Evaluation:", eval_res)
```
