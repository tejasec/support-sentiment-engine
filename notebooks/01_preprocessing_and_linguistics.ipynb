# Text Processing & Linguistic Foundations

In this notebook, we implement and evaluate:
1. Text Ingestion & Modular Cleaning (case folding, regex noise removal, URLs, punctuation)
2. Word, Sentence, and Subword Byte-Pair Encoding (BPE) tokenization
3. Morphological Processing: Comparative study of Porter Stemming vs. WordNet Lemmatization
4. Linguistic Ambiguity Detection (Lexical and Syntactic disambiguation)


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from nlp_engine.preprocessing.cleaner import TextCleaner
from nlp_engine.preprocessing.tokenizer import WordTokenizer, SentenceTokenizer, BytePairEncoder
from nlp_engine.preprocessing.morphology import MorphologyProcessor
from nlp_engine.preprocessing.ambiguity import AmbiguityHandler
```

## 1. Text Normalization Pipeline


```python
cleaner = TextCleaner(lowercase=True, strip_punctuation=True, remove_urls=True, remove_emails=True)
sample_ticket = "URGENT: Billed $49.99 twice for order ORD-99120 on 2026-04-12! Visit http://support.xyz/help or email support@cloud.io."
cleaned_text = cleaner.clean(sample_ticket)
print("Raw Text:    ", sample_ticket)
print("Cleaned Text:", cleaned_text)
```

## 2. Tokenization & Subword BPE


```python
word_tok = WordTokenizer()
sent_tok = SentenceTokenizer()
tokens = word_tok.tokenize(cleaned_text)
print("Word Tokens:", tokens)

bpe = BytePairEncoder(num_merges=25)
bpe.fit(["order", "ordering", "orders", "charged", "charging", "charge"])
print("BPE Subwords for 'ordering charges':", bpe.tokenize("ordering charges"))
```

## 3. Morphological Processing: Stemming vs. Lemmatization


```python
morph = MorphologyProcessor()
test_words = ["running", "orders", "deliveries", "charged", "cancellation"]
comparison = morph.compare_stem_vs_lemma(test_words)

import pandas as pd
pd.DataFrame(comparison)
```

## 4. Linguistic Ambiguity Resolution


```python
ambig = AmbiguityHandler()
print("Sense of 'charge' in credit card context:", ambig.disambiguate_word("charge", ["credit", "card", "refund", "fee"])) 
print("Sense of 'charge' in battery context:    ", ambig.disambiguate_word("charge", ["battery", "phone", "cable"]))
```
